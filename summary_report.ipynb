# Summary Report

## Business Problem
Fraud detection is critical in financial systems. Standard accuracy fails because the dataset is highly imbalanced — fraud cases are less than 0.2%. A model predicting only “non‑fraud” would show >99% accuracy but miss all fraud, leading to huge financial losses.

## Imbalance Handling Strategy
We tested multiple strategies: class weighting, SMOTE oversampling, and undersampling. SMOTE improved recall significantly, but the tuned XGBoost model handled imbalance most effectively, delivering the highest PR‑AUC.

## Recommended Model & Threshold
Tuned XGBoost achieved PR‑AUC ≈ 0.89. Threshold tuning showed two options:  
- F1‑optimal (~0.97) → best ML balance.  
- Recall ≥0.90 (~0.25) → maximum fraud coverage.  
We recommend Recall ≥0.90 threshold because missing fraud is costlier than investigating extra cases.

## Cost-Benefit Result
On 50,000 transactions (≈85 fraud cases):  
- TP ≈ 77 → Money Saved = ₹3.46 lakh  
- FP ≈ 200 → Investigation Cost = ₹41,550  
- FN ≈ 8 → Money Lost = ₹36,000  
- **Net Benefit ≈ ₹3.04 lakh**  

This shows business‑optimal threshold differs from ML‑optimal threshold.

## Production Improvements
- Real‑time scoring pipeline for instant alerts.  
- Model drift monitoring to adapt to new fraud patterns.  
- Feedback loop with analysts for retraining.  
- Threshold tuning as a business decision, not just ML metric.
